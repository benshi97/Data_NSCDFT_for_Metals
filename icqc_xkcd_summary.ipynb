```python
import matplotlib.pyplot as plt
from jup_plot import *

# Hand-drawn sketch effect without needing an xkcd font
plt.rcParams["path.sketch"] = (1, 120, 2)

labels = ["SOTA\n(i.e., RPA)", "This work\ndhBEEF-vdW"]
values = [20, 12]

fig, ax = plt.subplots(figsize=(2,2), dpi=600)

# Bars
bars = ax.bar(labels, values, width=0.55, linewidth=2,color=['red','green'])

# Transition-metal chemical accuracy as a filled band
ax.axhspan(0, 13, alpha=0.15)
ax.axhline(13, linestyle="--", linewidth=2)

# Annotation for the band
ax.text(
    0.3, 13,
    "Transition-Metal\nChemical Accuracy",
    fontsize=8,
    va="bottom"
)



ax.set_ylabel("Error [kJ/mol]")
ax.set_ylim(0, 22)

# Clean up the look
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("dft_comparison.png", dpi=600,transparent=True)
```
